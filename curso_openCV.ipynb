```python
import cv2
```

```python
image = cv2.imread('perro.jpg')
```

```python
type(image)
```

```python
plt.imshow(image)
```

```python
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image)
```

```python
image_black_white = cv2.imread('perro.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
image_black_white
```

```python
plt.imshow(image_black_white, cmap = 'gray')
```

```python
image2 = cv2.resize(image, (600,500))
```

```python
plt.imshow(image2)
```

```python
ratio_wide = 0.5
ratio_length =0.5
```

```python
image2 = cv2.resize(image, (0,0), image, ratio_wide, ratio_length)
```

```python
plt.imshow(image2)
```

```python
image2.shape
```

```python
image2_flip = cv2.flip(image2, 1)
```

```python
plt.imshow(image2_flip)
```

```python
image2_flip = cv2.flip(image2, 0)
```

```python
plt.imshow(image2_flip)
```

```python
black = np.zeros(shape=(500,500,3))
```

```python
plt.imshow(black)
```

```python
cv2.rectangle(black, pt1=(20,20), pt2=(100,100), color=(255,12,0), thickness=5)
plt.imshow(black)
```

```python
cv2.circle(black, center=(250,250), radius=100, color=(0,0,255), thickness=5)
plt.imshow(black)
```

```python
cv2.line(black, pt1=(0,400), pt2=(500,400), color=(0,255,0), thickness=(5))
```

```python
plt.imshow(black)
```

```python
black = np.zeros(shape=(500,500,3), dtype=np.int16)
```

```python
plt.imshow(black)
```

```python
fuente = cv2.FONT_HERSHEY_SCRIPT_COMPLEX
cv2.putText(black, text='hello', org=(100,250), fontFace=fuente, fontScale=3, color=(0,255,0), thickness=4, lineType=cv2.LINE_8)
```

```python
plt.imshow(black);
```

```python
vertex_points = np.array([[300,300],[400,200],[100,250],[250,450]])
vertex_points
```

```python
points = vertex_points.reshape(-1,1,2)
points.shape
```

```python
cv2.polylines(black,[points], isClosed=True, color=(255,255,255), thickness=5)
plt.imshow(black)
```
